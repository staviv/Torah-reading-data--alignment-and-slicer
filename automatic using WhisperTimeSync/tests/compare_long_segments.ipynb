# השוואת קבצי כתוביות SRT

קוד זה מיועד להשוואה בין שני קבצי כתוביות SRT:
1. קובץ המכיל את החלוקה לזמנים שיצר מודל הבינה המלאכותית 
2. קובץ לאחר התאמה והתיקון למילים הנכונות

## הרעיון
אם המילה הראשונה זהה גם בקטע הנוכחי וגם בקטע הבא, סימן שהקטע הותאם בצורה נכונה.
לכן כל שעלינו לעשות הוא להשוות את המילה הראשונה בכל קטע ולהציג את ההבדלים.
כך נוכל להשוות בקלות בין הקבצים ולזהות תקלות.

## תיאור הפונקציונליות

הקוד מבצע את הפעולות הבאות:
- קורא שני קבצי SRT ומנקה אותם מניקוד וטעמים
- משווה את המילה הראשונה בכל כתובית בין שני הקבצים
- מציג את ההבדלים שנמצאו: מספר הכתובית והמילים השונות בכל קובץ
- מאפשר שמירת התוצאות לקובץ SRT חדש

```python
import srt
from datetime import timedelta
from typing import List, Tuple
import sys
sys.path.append('..')
from nikud_and_teamim import remove_nikud_and_teamim

def read_srt_file(filename: str) -> List[srt.Subtitle]:
    """Read and parse SRT file"""
    with open(filename, 'r', encoding='utf-8') as f:
        clean_text = remove_nikud_and_teamim(f.read()).replace("־", " ").replace("׃", "")
        return list(srt.parse(clean_text))

def compare_first_words(file1: str, file2: str) -> List[Tuple[int, str, str]]:
    """Compare first words of subtitles between two SRT files"""
    subs1 = read_srt_file(file1)
    subs2 = read_srt_file(file2)
    
    differences = []
    for i, (s1, s2) in enumerate(zip(subs1, subs2), 1):
        word1 = s1.content.split()[0] if s1.content.split() else ''
        word2 = s2.content.split()[0] if s2.content.split() else ''
        
        if word1 != word2:
            differences.append((i, word1, word2))
    
    return differences

# Compare files and display results
differences = compare_first_words('../final_step.txt.srt', '../output.srt')

for idx, word1, word2 in differences:
    print(f"Subtitle {idx}:")
    print(f"File 1 first word: '{word1}'")
    print(f"File 2 first word: '{word2}'")
    print("-" * 40)

# If you need to write results to a new SRT file:
def write_srt_file(subtitles: List[srt.Subtitle], filename: str):
    """Write subtitles to SRT file"""
    with open(filename, 'w', encoding='utf-8') as f:
        f.write(srt.compose(subtitles))
```
